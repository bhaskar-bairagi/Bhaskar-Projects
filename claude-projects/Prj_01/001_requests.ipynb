```python
# Create an API client
from anthropic import Anthropic

client = Anthropic()
model = "claude-sonnet-5"
```

```python
def add_user_message(messages, text):        # Defining helper functions for user messsage
    user_message = {"role": "user", "content": text}
    messages.append (user_message)

def add_assistant_message(messages, text):      # Defining helper functions for assistant's messsage
    assistant_message = {"role": "assistant", "content": text}
    messages.append (assistant_message)

def chat(messages):       # Defining helper functions for chat messsage 
    message= client.messages.create(
        model= model, 
        max_tokens=1000, 
        messages= messages,
    )
    return message.content[0].text
"""
    
            {
            "role": "user",
            "content": "What is quantum computing? Answer in one sentence"
        }
    ]
)
# Make another request
message = client.messages.create(
    model= model, 
    max_tokens=1000, 
    messages=[
        {
            "role": "user",
            "content": "Write another sentence"
        }
    ]
)  """
```
